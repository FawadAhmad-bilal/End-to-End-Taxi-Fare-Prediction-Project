```python
df=df.dropna()
```

```python
df.drop(["pickup","dropoff","fare","tip","tolls"],axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

# ── 1. BINARY COLUMNS ──────────────────────────
le = LabelEncoder()
df['color_encoded']   = le.fit_transform(df['color'])
df['payment_encoded'] = le.fit_transform(df['payment'])

# ── 2. HIGH CARDINALITY (pickup_zone, dropoff_zone) ─────────────
for col in ['pickup_zone', 'dropoff_zone']:
    freq_map = df[col].value_counts().to_dict()
    df[f'{col}_encoded'] = df[col].map(freq_map).astype(int)  # ✅ yahan lagao

# ── 3. LOW CARDINALITY (borough columns) ────────────────────────
df = pd.get_dummies(df, columns=['pickup_borough', 'dropoff_borough'],
                    drop_first=True, dtype=int)  # ✅ dtype=int yahan lagao

# ── 4. ORIGINAL COLUMNS DROP KARO ───────────────────────────────
df.drop(columns=['color', 'payment', 'pickup_zone', 'dropoff_zone'],
        inplace=True)

print(df.shape)
print(df.columns.tolist())
```

```python
df.head()
```

```python
plt.figure(figsize=(16,6))
sns.boxplot(df)
```

```python
x=df.drop("total",axis=1)
y=df["total"]
```

```python
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
from sklearn.preprocessing import StandardScaler
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
sc=StandardScaler()
```

```python
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)
```

```python
ml=LinearRegression()
```

```python
ml.fit(x_train_sc,y_train)
```

```python
ml.score(x_train_sc,y_train)
```

```python
pre=ml.predict(x_test_sc)
```

```python
r2_score(y_test,pre)
```

```python
mean_squared_error(y_test,pre)
```

```python
mean_absolute_error(y_test,pre)
```

```python
from sklearn.linear_model import Lasso
```

```python
model=Lasso(alpha=1)
```

```python
model.fit(x_train_sc,y_train)
```

```python
model.score(x_train_sc,y_train)
```

```python
pre2=model.predict(x_test_sc)
```

```python
r2_score(y_test,pre2)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
RandomForestRegressor().get_params()
```

```python
ml2=RandomForestRegressor(max_depth= 5, min_samples_leaf=1, min_samples_split= 4,n_estimators=200)
```

```python
ml2.fit(x_train,y_train)
```

```python
ml2.score(x_train,y_train)
```

```python
pre2=ml2.predict(x_test)
```

```python
r2_score(y_test,pre2)
```
